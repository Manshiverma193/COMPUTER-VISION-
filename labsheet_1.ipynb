2. Write a program to Convert a color image into Grayscale and Binary Image.

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get image filename
filename = next(iter(uploaded))
print("Image Name:", filename)

# Read the color image
image = cv2.imread(filename)

# Convert color image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Convert grayscale image to binary image
_, binary_image = cv2.threshold(gray_image, 100, 255, cv2.THRESH_BINARY)

# Display original color image
print("Original Color Image:")
cv2_imshow(image)

# Display grayscale image
print("Grayscale Image:")
cv2_imshow(gray_image)

# Display binary image
print("Binary Image:")
cv2_imshow(binary_image)

# Save images
cv2.imwrite("grayscale.jpg", gray_image)
cv2.imwrite("binary.jpg", binary_image)

print("Grayscale and Binary images saved successfully.")
```

3. Write a Program to compute and Display the complement(Negative) of an Image.

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get image filename
filename = next(iter(uploaded))

# Read the original color image
image = cv2.imread(filename)

# Convert color image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Compute complement (negative) of color image
negative_color = 255 - image

# Compute complement (negative) of grayscale image
negative_gray = 255 - gray_image

# Display Original Color Image
print("Original Color Image:")
cv2_imshow(image)

# Display Grayscale Image
print("Grayscale Image:")
cv2_imshow(gray_image)

# Display Negative Color Image
print("Negative (Complement) Color Image:")
cv2_imshow(negative_color)

# Display Negative Grayscale Image
print("Negative (Complement) Grayscale Image:")
cv2_imshow(negative_gray)
```

4. Write a Program to Resize ,Crop and display an image.

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get image filename
filename = next(iter(uploaded))

# Read the image
image = cv2.imread(filename)

# Display original image
print("Original Image:")
cv2_imshow(image)

# Resize the image
resized_image = cv2.resize(image, (300, 300))

print("Resized Image:")
cv2_imshow(resized_image)

# Crop the image
# Format: image[y1:y2, x1:x2]
cropped_image = image[50:300, 50:300]

print("Cropped Image:")
cv2_imshow(cropped_image)
```

5. Write a Python program using OpenCV to split an image into two parts and merge the parts back into the original image

```python
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read image
image = cv2.imread(filename)

# Get image dimensions
height, width = image.shape[:2]

# Split image into two parts
part1 = image[:, :width//2]
part2 = image[:, width//2:]

# Display split images
print("First Part:")
cv2_imshow(part1)

print("Second Part:")
cv2_imshow(part2)

# Merge the two parts
merged_image = cv2.hconcat([part1, part2])

# Display merged image
print("Merged Image:")
cv2_imshow(merged_image)

# Save merged image
cv2.imwrite("merged.jpg", merged_image)
```
